```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 200)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x, y, linewidth=2, label="sin(x)")
ax.set(title="2D Line Plot", xlabel="x", ylabel="sin(x)")
ax.grid(alpha=0.3)
ax.legend()
fig.tight_layout()
plt.show()

# color, linestyle, linewidth, marker, markersize, alpha, label.
```

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
x = rng.normal(size=100)
y = 2 * x + rng.normal(scale=0.7, size=100)

fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(x, y, s=45, alpha=0.7, edgecolors="none")
ax.set(title="Scatter Plot", xlabel="X", ylabel="Y")
fig.tight_layout()
plt.show()

# s for marker area, c for color values, alpha, marker, edgecolors, cmap.
```

```python
import matplotlib.pyplot as plt

labels = ["A", "B", "C", "D"]
values = [24, 18, 30, 22]

fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(labels, values)
ax.set(title="Bar Chart", xlabel="Category", ylabel="Value")
fig.tight_layout()
plt.show()

# width, color, edgecolor, alpha, label, align.
```

```python
import numpy as np
import matplotlib.pyplot as plt

labels = ["Q1", "Q2", "Q3", "Q4"]
product_a = np.array([20, 25, 22, 30])
product_b = np.array([15, 18, 20, 17])

fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(labels, product_a, label="Product A")
ax.bar(labels, product_b, bottom=product_a, label="Product B")
ax.set(title="Stacked Bar Chart", ylabel="Value")
ax.legend()
fig.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
x = rng.normal(size=1000)

fig, ax = plt.subplots(figsize=(8, 5))
ax.hist(x, bins=30, alpha=0.8, edgecolor="black")
ax.set(title="Histogram", xlabel="Value", ylabel="Frequency")
fig.tight_layout()
plt.show()

# bins, density, range, alpha, histtype, orientation, edgecolor.
```

```python
import matplotlib.pyplot as plt

labels = ["Python", "C++", "Java", "Other"]
values = [55, 20, 15, 10]

fig, ax = plt.subplots(figsize=(6, 6))
ax.pie(values, labels=labels, autopct="%1.1f%%", startangle=90)
ax.set_title("Pie Chart")
fig.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 200)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x, y)
ax.axhline(0, linestyle="--", alpha=0.6)
ax.axvline(np.pi, linestyle="--", alpha=0.6)
ax.annotate("Peak", xy=(np.pi / 2, 1), xytext=(4, 1.25),
            arrowprops={"arrowstyle": "->"})
ax.text(0.5, -0.8, "Reference region")
fig.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 200)

fig, axes = plt.subplots(2, 2, figsize=(10, 7), sharex=True)
axes[0, 0].plot(x, np.sin(x))
axes[0, 1].plot(x, np.cos(x))
axes[1, 0].plot(x, np.sin(2 * x))
axes[1, 1].plot(x, np.cos(2 * x))
fig.suptitle("Subplots")
fig.tight_layout()
plt.show()

# nrows, ncols, figsize, sharex, sharey, squeeze, constrained_layout.
```

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")

x = rng.normal(size=100)
y = rng.normal(size=100)
z = x + y + rng.normal(scale=0.5, size=100)

ax.scatter(x, y, z, s=35, alpha=0.7)
ax.set(xlabel="X", ylabel="Y", zlabel="Z", title="3D Scatter")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 200)
y = np.sin(x)
z = np.cos(x)

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")
ax.plot(x, y, z, linewidth=2)
ax.set(xlabel="X", ylabel="Y", zlabel="Z", title="3D Line Plot")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")
ax.plot_surface(X, Y, Z, cmap="viridis")
ax.set(title="3D Surface Plot", xlabel="X", ylabel="Y", zlabel="Z")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

fig, ax = plt.subplots(figsize=(8, 6))
contours = ax.contour(X, Y, Z, levels=12)
ax.clabel(contours, inline=True, fontsize=8)
ax.set(title="Contour Plot", xlabel="X", ylabel="Y")
fig.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
data = rng.normal(size=(8, 10))

fig, ax = plt.subplots(figsize=(9, 5))
im = ax.imshow(data, cmap="coolwarm", aspect="auto")
fig.colorbar(im, ax=ax, label="Value")
ax.set(title="Heat Map", xlabel="Column", ylabel="Row")
fig.tight_layout()
plt.show()
```
